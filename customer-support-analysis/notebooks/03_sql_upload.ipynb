```python
import pandas as pd
import sqlite3

# Paths to your files
csv_file_path = "../data/predicted_support_tickets.csv"
sqlite_db_path = "../data/customer_tickets.db"

# Load the predicted CSV
df = pd.read_csv(csv_file_path)
print("CSV loaded successfully. Shape:", df.shape)

# Connect to the SQLite DB (creates if not exists)
conn = sqlite3.connect(sqlite_db_path)

# Save to a new table
df.to_sql("predicted_support_tickets", conn, if_exists="replace", index=False)
print("Table 'predicted_support_tickets' created/replaced in database.")

# # Optional: Verify
# tables = pd.read_sql_query("SELECT name FROM sqlite_master WHERE type='table';", conn)
# print("Tables in database:", tables)

conn.close()

print("Uploaded predicted_support_tickets to DB")
```
